# SUPPLY CHAIN DATA ANALYSIS PYTHON PROJECT

**Business problem** - The organization manages a multi-country supply chain involving numerous suppliers, warehouses, carriers, and shipping modes. Operational data is captured across systems but is rarely analyzed in a structured, consistent way. As a result, leadership lacks clear visibility. Without a structured analytics process, decisions on supplier selection, carrier contracts and inventory planning are made reactively rather than being backed by data.

**Solution** - The solution uses Python-based data analysis with NumPy, Pandas, Matplotlib, and Seaborn to transform raw supply chain data into structured, actionable insights. The analysis includes data loading and inspection, data cleaning and validation, exploratory analysis, feature engineering, Univariate analysis, Bivariate analysis, Time series analysis, KPI calculation and answering important business questions to evaluate suppliers, warehouses, carriers, shipping modes, products, inventory, orders and delivery performance and present the final insights.

## 4. Feature Engineering

```python
# Calculate delivery time
df['delivery_time'] = (df['delivery_date'] - df['order_date']).dt.days
```

```python
df[['delivery_date','order_date','delivery_time']].dropna()
```

```python
df['delivery_delay_days'] = df['delivery_time'] - df['lead_time_days']
```

```python
df[['delivery_time','lead_time_days','delivery_delay_days']]
df['delivery_time'] = df['delivery_time'].fillna(0)
df['delivery_delay_days'] = df['delivery_delay_days'].fillna(0)
```

```python
df[['delivery_time','lead_time_days','delivery_delay_days']]
```

```python
df['delivery_status'] = np.select(
    [df['delivery_delay_days']>0,df['delivery_delay_days']==0,df['delivery_delay_days']<0],
    ['Late','On Time','Early'],
 default='Unknown')
```

```python
df[['delivery_time','lead_time_days','delivery_delay_days','delivery_status']]
```

```python
# Calculate cost per unit

df['cost_per_unit'] = np.where(df['quantity']>0,df['total_cost']/df['quantity'],np.nan)
```

```python
df[['quantity','total_cost','cost_per_unit']]
```

```python
# Extract useful time features from order date
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_month_name'] = df['order_date'].dt.month_name().str[:3]
df['order_quarter'] = 'Q'+df['order_date'].dt.quarter.astype(str)
df['order_day'] = df['order_date'].dt.day_name().str[:3]
```

```python
df[['order_date','order_year','order_month','order_month_name','order_quarter','order_day']]
```

```python
df['order_quarter'] = df['order_quarter'].replace('Qnan','')
```

```python
# Monthly sales trend

df['sales'] = df['quantity'] * df['unit_cost']
monthly_trend = (
    df.groupby(['order_month', 'order_month_name'])['sales']
      .sum()
      .reset_index()
      .sort_index()
)
monthly_trend
```

```python
sns.relplot(data=monthly_trend,x='order_month_name',y='sales', kind='line', marker='o',color=husl_colors[11],markerfacecolor='blue',linewidth=2)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales (In Billions)')
plt.savefig('Fig_12_monthlysales.png')
plt.show()
```

★ Insight: April and July recorded the highest sales, while February had the lowest sales, indicating significant variation in monthly sales performance.

```python
# Monthly Order Trend
order_trend = (
    df.groupby(['order_month', 'order_month_name'])['order_id']
      .nunique()
      .reset_index(name='total_orders')
      .sort_values('order_month')
)

order_trend
```

```python

total_orders = df['order_id'].nunique()
sns.relplot(data=order_trend,x='order_month_name',y='total_orders', kind='line', marker='o',color=husl_colors[12],markerfacecolor='m',linewidth=2)
plt.title('Monthly Orders Trend')
plt.xlabel('Month')
plt.ylabel('Orders')
plt.savefig('Fig_13_monthlyorders.png')
plt.show()
```
